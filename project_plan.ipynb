# Projeto Final — Decomposição

## Análise de Teste A/B do Sistema de Recomendação

Este projeto tem como objetivo avaliar os resultados de um teste A/B realizado para comparar o sistema de recomendação atual com uma versão aprimorada.

A análise será utilizada para apoiar a decisão da área de Produto sobre a implantação do novo sistema de recomendação.

## 1. Problema de negócio

A área de Produto precisa decidir se a versão aprimorada do sistema de recomendação deve substituir o sistema atualmente utilizado.

Para apoiar essa decisão, será necessário comparar o comportamento dos usuários dos dois grupos do teste A/B e verificar se a nova versão apresenta melhora nas principais etapas de conversão.

A análise deverá identificar se existe diferença relevante entre os grupos e se os resultados observados são suficientes para recomendar a implantação do novo sistema.

## 2. Pergunta de negócio

A nova versão do sistema de recomendação apresenta desempenho superior à versão atual e, portanto, deve ser implantada para todos os usuários?

## 3. Objetivos da análise

- Comparar o desempenho dos grupos A e B no teste A/B.
- Calcular as principais taxas de conversão em cada etapa da jornada do usuário.
- Verificar se existem diferenças estatisticamente significativas entre os grupos.
- Identificar em quais etapas a nova versão apresenta maior ou menor desempenho.
- Avaliar a variação relativa do grupo B em relação ao grupo A.
- Fornecer uma recomendação para a área de Produto sobre a implantação do novo sistema.

## 4. Indicadores (KPIs)

Para responder à pergunta de negócio, serão analisados os seguintes indicadores:

### 4.1 Taxa de conversão por etapa

**Fórmula:** número de usuários do grupo que realizaram o evento da etapa dividido pelo número de usuários do grupo que realizaram login.

**Objetivo:** medir o percentual de usuários que avançam em cada etapa da jornada e comparar o desempenho dos grupos A e B no teste A/B.

---

### 4.2 Variação relativa de B em relação a A

**Fórmula:** (taxa de conversão do grupo B − taxa de conversão do grupo A) dividido pela taxa de conversão do grupo A.

**Objetivo:** medir o quanto o desempenho do grupo B aumentou ou diminuiu em relação ao grupo A e verificar se a variação observada atingiu a meta de aumento de 10%.


## 5. Dados utilizados

A análise será realizada utilizando quatro conjuntos de dados:

- `ab_project_marketing_events_us.csv`: calendário dos eventos de marketing realizados durante 2020.
- `final_ab_new_users_upd_us.csv`: informações dos usuários que se cadastraram entre 7 e 21 de dezembro de 2020, incluindo `user_id`, data de cadastro, região e dispositivo.
- `final_ab_events_upd_us.csv`: registros dos eventos realizados pelos novos usuários entre 7 de dezembro de 2020 e 1º de janeiro de 2021.
- `final_ab_participants_upd_us.csv`: informações sobre os usuários participantes do teste, incluindo `user_id`, teste e grupo experimental.

A chave principal para relacionar as bases será `user_id`.

As bases serão utilizadas para identificar os participantes do experimento, acompanhar seus eventos e comparar a conversão dos grupos A e B.

## 6. Pré-processamento dos dados

Antes da análise, os dados serão verificados e preparados para garantir sua qualidade e consistência.

Serão realizadas as seguintes etapas:

- Verificar a estrutura e os tipos de dados de cada tabela.
- Converter as colunas de data para o formato adequado.
- Identificar e caracterizar valores ausentes.
- Identificar registros duplicados.
- Verificar a consistência da chave `user_id` entre as bases.
- Verificar se existem usuários participantes de mais de um teste A/B.
- Avaliar a distribuição dos usuários entre os grupos A e B.
- Verificar se existem usuários ou eventos fora do período definido para o experimento.
- Integrar as bases por meio de `user_id`, quando necessário para a análise.

Um cuidado importante será investigar a possível sobreposição de usuários entre os testes. Usuários participantes de mais de um experimento podem interferir na independência dos grupos e afetar a interpretação dos resultados.

Os tratamentos aplicados serão documentados e realizados no próprio notebook para que o processo seja reproduzível.

## 7. Análise Exploratória de Dados

A análise exploratória terá como objetivo avaliar a composição dos grupos, o comportamento dos usuários e a qualidade dos dados antes da realização do teste estatístico.

Serão realizadas as seguintes análises:

- Comparar a quantidade de usuários nos grupos A e B.
- Verificar se os usuários estão distribuídos adequadamente entre os grupos.
- Analisar a quantidade de eventos realizados por usuário em cada grupo.
- Verificar se o número de eventos por usuário apresenta distribuição semelhante entre as amostras.
- Analisar a presença dos usuários em cada grupo e verificar possíveis sobreposições entre os testes.
- Avaliar a distribuição dos eventos ao longo dos dias do experimento.
- Analisar o funil de conversão `product_page → product_cart → purchase`.
- Calcular e comparar a conversão de cada etapa entre os grupos A e B.
- Identificar possíveis particularidades ou anomalias nos dados que possam afetar a validade do teste A/B.

As visualizações serão utilizadas para facilitar a comparação entre os grupos e identificar padrões ou desvios relevantes.

Ao final da AED, serão definidos os principais pontos que devem ser considerados antes da aplicação dos testes estatísticos.

## 8. Teste de hipóteses

Serão realizados testes estatísticos para verificar se as diferenças de conversão entre os grupos A e B são significativas.

Para cada etapa do funil (`product_page`, `product_cart` e `purchase`), serão consideradas:

- **H0:** a taxa de conversão é igual nos grupos A e B.
- **H1:** a taxa de conversão é diferente entre os grupos A e B.

Será utilizado o **teste z para comparação de duas proporções**, com nível de significância de 5%. O p-valor de cada comparação será analisado em relação a esse nível de significância.

Como serão realizadas três comparações, os resultados também deverão ser interpretados considerando a possibilidade de aumento do erro do tipo I em testes múltiplos.

A decisão será baseada no p-valor:

- se `p-valor < 0,05`, rejeitar H0;
- se `p-valor ≥ 0,05`, não rejeitar H0.

Os resultados serão comparados com a meta de aumento de **10% na conversão** estabelecida para cada etapa do funil.


## 9. Conclusões e recomendações

- Consolidar os resultados da análise exploratória e dos testes estatísticos.
- Comparar as taxas de conversão dos grupos A e B em cada etapa do funil.
- Verificar se o aumento de 10% foi atingido em cada etapa.
- Avaliar a significância estatística das diferenças observadas.
- Considerar possíveis limitações identificadas durante a análise dos dados e do experimento.
- Responder à pergunta de negócio e recomendar se o novo sistema de recomendação deve ser implantado.
