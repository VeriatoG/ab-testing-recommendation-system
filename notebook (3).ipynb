## 8. Teste de hipóteses

Serão realizados testes estatísticos para verificar se as diferenças de conversão entre os grupos A e B são significativas.

Para cada etapa do funil (`product_page`, `product_cart` e `purchase`), serão consideradas:

- **H0:** a taxa de conversão é igual nos grupos A e B.
- **H1:** a taxa de conversão é diferente entre os grupos A e B.

Será utilizado o **teste z para comparação de duas proporções**, com nível de significância de 5%.

Como serão realizadas três comparações, será aplicada a **correção de Bonferroni** para controlar o risco de erro estatístico causado pelos testes múltiplos.

A decisão será baseada no p-valor ajustado:

- se `p-valor < α ajustado`, rejeitar H0;
- se `p-valor ≥ α ajustado`, não rejeitar H0.

Os resultados serão comparados com a meta de aumento de **10% na conversão** estabelecida para cada etapa do funil.
